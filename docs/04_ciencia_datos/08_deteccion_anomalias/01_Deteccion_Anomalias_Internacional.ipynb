# Deteccion de anomalias en trafico internacional
Esta subfase implementa un baseline estadistico y Isolation Forest sobre series de velocidad a 5 minutos. El objetivo es detectar observaciones inusuales y comparar estabilidad por fuente; sin etiquetas externas no se interpreta como deteccion confirmada de incidentes.

```python
from pathlib import Path
import json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.ensemble import IsolationForest
from sklearn.preprocessing import StandardScaler

PROJECT_ROOT = Path.cwd()
while not (PROJECT_ROOT / 'data').exists() and PROJECT_ROOT != PROJECT_ROOT.parent:
    PROJECT_ROOT = PROJECT_ROOT.parent
OUTPUT_DIR = PROJECT_ROOT / 'data' / '04_model_output' / 'anomalies'
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
SOURCE_PATHS = {
    'metr_la': PROJECT_ROOT / 'data' / '01_processed' / 'external' / 'metr_la.parquet',
    'pems_bay': PROJECT_ROOT / 'data' / '01_processed' / 'external' / 'pems_bay.parquet',
}
print('Fuentes:', list(SOURCE_PATHS))
```

```python
def build_features(source, path):
    raw = pd.read_parquet(path)
    raw['timestamp'] = pd.to_datetime(raw['timestamp'], errors='coerce')
    sensors = [column for column in raw.columns if column != 'timestamp']
    frame = pd.DataFrame({'timestamp': raw['timestamp'], 'speed_median': raw[sensors].apply(pd.to_numeric, errors='coerce').median(axis=1)})
    frame = frame.dropna().sort_values('timestamp').reset_index(drop=True)
    frame['source'] = source
    frame['lag_1'] = frame['speed_median'].shift(1)
    frame['delta_1'] = frame['speed_median'] - frame['lag_1']
    frame['rolling_median_3'] = frame['speed_median'].shift(1).rolling(3, min_periods=3).median()
    frame['rolling_mad_3'] = frame['speed_median'].shift(1).rolling(3, min_periods=3).apply(lambda values: np.median(np.abs(values - np.median(values))))
    frame['hour'] = frame['timestamp'].dt.hour
    frame['day_of_week'] = frame['timestamp'].dt.dayofweek
    return frame.dropna().reset_index(drop=True)

features = pd.concat([build_features(source, path) for source, path in SOURCE_PATHS.items()], ignore_index=True)
display(features.groupby('source').agg(filas=('timestamp', 'size'), inicio=('timestamp', 'min'), fin=('timestamp', 'max')))
print('filas_features=', len(features))
```

```python
features['robust_z_speed'] = (features['speed_median'] - features['rolling_median_3']) / (1.4826 * features['rolling_mad_3'].replace(0, np.nan))
features['baseline_anomaly'] = (features['robust_z_speed'].abs() >= 3).fillna(False)
MODEL_FEATURES = ['speed_median', 'delta_1', 'rolling_median_3', 'rolling_mad_3', 'hour', 'day_of_week']
anomaly_rows = []
for source, group in features.groupby('source', sort=False):
    model = IsolationForest(n_estimators=150, contamination=0.05, random_state=42, n_jobs=-1)
    scaled = StandardScaler().fit_transform(group[MODEL_FEATURES])
    isolation_prediction = model.fit_predict(scaled)
    source_result = group.copy()
    source_result['isolation_anomaly'] = isolation_prediction == -1
    source_result['agreement'] = source_result['baseline_anomaly'] & source_result['isolation_anomaly']
    anomaly_rows.append(source_result)
anomalies = pd.concat(anomaly_rows, ignore_index=True)
display(anomalies.groupby('source', as_index=False).agg(
    observaciones=('timestamp', 'size'),
    baseline_anomalias=('baseline_anomaly', 'sum'),
    isolation_anomalias=('isolation_anomaly', 'sum'),
    coincidencias=('agreement', 'sum'),
))
```

```python
summary = anomalies.groupby('source', as_index=False).agg(
    observaciones=('timestamp', 'size'),
    baseline_anomalias=('baseline_anomaly', 'sum'),
    isolation_anomalias=('isolation_anomaly', 'sum'),
    coincidencias=('agreement', 'sum'),
)
summary['tasa_isolation_pct'] = summary['isolation_anomalias'] / summary['observaciones'] * 100
summary['tasa_coincidencia_sobre_baseline_pct'] = summary['coincidencias'] / summary['baseline_anomalias'].replace(0, np.nan) * 100
display(summary)
anomalies.to_parquet(OUTPUT_DIR / 'anomalias_alta_frecuencia.parquet', index=False)
summary.to_csv(OUTPUT_DIR / 'resumen_anomalias.csv', index=False)
```

```python
sample_parts = []
for source, group in anomalies.groupby('source', sort=False):
    sample_parts.append(group.sort_values('timestamp').tail(1000))
sample = pd.concat(sample_parts, ignore_index=True)
fig, axes = plt.subplots(1, 2, figsize=(15, 5))
for source, group in sample.groupby('source'):
    axes[0].plot(group['timestamp'], group['speed_median'], linewidth=1, label=source)
    flagged = group[group['isolation_anomaly']]
    axes[0].scatter(flagged['timestamp'], flagged['speed_median'], s=12)
axes[0].set_title('Velocidad y anomalías Isolation Forest')
axes[0].set_ylabel('Velocidad mediana')
axes[0].legend(fontsize=8)
summary.set_index('source')[['baseline_anomalias', 'isolation_anomalias', 'coincidencias']].plot.bar(ax=axes[1])
axes[1].set_title('Comparación de métodos')
axes[1].set_ylabel('Cantidad')
plt.tight_layout()
plt.show()
```

```python
report = {
    'experiment': 'deteccion_anomalias_alta_frecuencia',
    'sources': list(SOURCE_PATHS),
    'resolution_minutes': 5,
    'methods': ['baseline_robust_z_score', 'isolation_forest'],
    'isolation_contamination': 0.05,
    'outputs': ['anomalias_alta_frecuencia.parquet', 'resumen_anomalias.csv'],
    'status': 'exploratory_anomaly_detection_without_ground_truth',
    'limitations': ['no_incident_ground_truth', 'network_median_proxy', 'highway_sources', 'not_Guadalajara_validation'],
}
(OUTPUT_DIR / 'resumen_experimento_anomalias.json').write_text(json.dumps(report, indent=2, ensure_ascii=False), encoding='utf-8')
print(json.dumps(report, indent=2, ensure_ascii=False))
```

## Interpretacion
Isolation Forest y el baseline estadistico producen candidatos de anomalía. La coincidencia entre ambos métodos es una medida de consistencia interna, no una precision contra incidentes reales. La etiqueta de incidente requiere validacion externa y revision de candidatos.
