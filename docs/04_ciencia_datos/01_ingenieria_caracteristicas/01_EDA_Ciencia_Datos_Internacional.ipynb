# EDA para ciencia de datos internacional
Este notebook inicia la fase de modelado con el TSI diario por fuente. La estratificacion `alto_tsi` es exploratoria y no sustituye una etiqueta operacional de deterioro.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import fisher_exact, ttest_ind

PROJECT_ROOT = Path.cwd()
while not (PROJECT_ROOT / 'data').exists() and PROJECT_ROOT != PROJECT_ROOT.parent:
    PROJECT_ROOT = PROJECT_ROOT.parent
INPUT = PROJECT_ROOT / 'data' / '03_analysis_output' / 'indicador_tsi' / 'tsi_provisional_diario.csv'
tsi = pd.read_csv(INPUT, parse_dates=['fecha'])
tsi['dia_semana'] = tsi['fecha'].dt.dayofweek
tsi['fin_de_semana'] = tsi['dia_semana'].isin([5, 6])
tsi['mes'] = tsi['fecha'].dt.month
tsi['trimestre'] = tsi['fecha'].dt.quarter
tsi.shape
```

```python
source_profile = tsi.groupby('fuente', as_index=False).agg(
    dias=('fecha', 'nunique'),
    observaciones=('observaciones', 'sum'),
    media_tsi=('tsi_provisional_media', 'mean'),
    desviacion_tsi=('tsi_provisional_media', 'std'),
    minimo_tsi=('tsi_provisional_media', 'min'),
    maximo_tsi=('tsi_provisional_media', 'max'),
)
source_profile['q75_tsi'] = tsi.groupby('fuente')['tsi_provisional_media'].quantile(0.75).values
display(source_profile.sort_values('dias', ascending=False))
```

```python
thresholds = tsi.groupby('fuente')['tsi_provisional_media'].transform('quantile', 0.75)
tsi['alto_tsi'] = tsi['tsi_provisional_media'] >= thresholds
tsi['grupo_dia'] = tsi['fin_de_semana'].map({False: 'laborable', True: 'fin_de_semana'})
display(tsi.groupby(['fuente', 'grupo_dia'], as_index=False).agg(
    dias=('fecha', 'nunique'),
    media_tsi=('tsi_provisional_media', 'mean'),
    alto_tsi=('alto_tsi', 'sum'),
))
```

```python
ab_rows = []
for source, group in tsi.groupby('fuente'):
    weekday = group.loc[~group['fin_de_semana'], 'tsi_provisional_media'].dropna()
    weekend = group.loc[group['fin_de_semana'], 'tsi_provisional_media'].dropna()
    if len(weekday) >= 2 and len(weekend) >= 2:
        test = ttest_ind(weekday, weekend, equal_var=False)
        pooled = ((len(weekday) - 1) * weekday.var() + (len(weekend) - 1) * weekend.var()) / (len(weekday) + len(weekend) - 2)
        effect = (weekday.mean() - weekend.mean()) / pooled ** 0.5 if pooled > 0 else 0.0
        ab_rows.append({'fuente': source, 'n_laborable': len(weekday), 'n_fin_de_semana': len(weekend), 'media_laborable': weekday.mean(), 'media_fin_de_semana': weekend.mean(), 'diferencia_media': weekday.mean() - weekend.mean(), 't_welch': test.statistic, 'p_valor': test.pvalue, 'd_cohen_aprox': effect})
ab_results = pd.DataFrame(ab_rows)
display(ab_results)
```

```python
fisher_rows = []
for source, group in tsi.groupby('fuente'):
    table = pd.crosstab(group['fin_de_semana'], group['alto_tsi']).reindex(index=[False, True], columns=[False, True], fill_value=0)
    odds_ratio, p_value = fisher_exact(table.to_numpy())
    fisher_rows.append({'fuente': source, 'laborable_bajo': int(table.loc[False, False]), 'laborable_alto': int(table.loc[False, True]), 'fin_semana_bajo': int(table.loc[True, False]), 'fin_semana_alto': int(table.loc[True, True]), 'odds_ratio': odds_ratio, 'p_valor': p_value})
fisher_results = pd.DataFrame(fisher_rows)
display(fisher_results)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(15, 5))
tsi.boxplot(column='tsi_provisional_media', by='grupo_dia', ax=axes[0], grid=False)
axes[0].set_title('TSI por tipo de dia')
axes[0].set_xlabel('Grupo')
axes[0].set_ylabel('TSI provisional')
tsi.groupby('fuente')['tsi_provisional_media'].mean().sort_values().plot.barh(ax=axes[1], color='#2f6690')
axes[1].set_title('Media TSI por fuente')
axes[1].set_xlabel('TSI provisional')
plt.suptitle('')
plt.tight_layout()
plt.show()
```
