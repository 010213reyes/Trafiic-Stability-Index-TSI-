```python
# Identificar el catálogo y esquema actuales
spark.sql("""
    SELECT
        current_catalog() AS catalogo_actual,
        current_schema() AS esquema_actual
""").show(truncate=False)
```

```python
# Crear el esquema exclusivo del proyecto TSI

spark.sql("""
    CREATE SCHEMA IF NOT EXISTS workspace.tsi
""")

print("Esquema TSI creado o existente.")
```

```python
spark.sql("""
    SHOW SCHEMAS IN workspace
""").show(truncate=False)
```

```python
spark.sql("""
    CREATE VOLUME IF NOT EXISTS workspace.tsi.datasets
""")

print("Volumen de almacenamiento TSI preparado.")
```

```python
spark.sql("""
    SHOW VOLUMES IN workspace.tsi
""").show(truncate=False)
```

```python
# Prueba de escritura en Unity Catalog

ruta = "/Volumes/workspace/tsi/datasets/prueba_tsi"

df_prueba = spark.createDataFrame(
    [
        ("Madrid", 65.5),
        ("Estambul", 48.2),
        ("Sao Paulo", 72.1)
    ],
    ["ciudad", "valor"]
)

df_prueba.write.mode("overwrite").parquet(ruta)

print("Archivo Parquet de prueba guardado correctamente.")
```

```python
# Verificar lectura del archivo Parquet

ruta = "/Volumes/workspace/tsi/datasets/prueba_tsi"

df_lectura = spark.read.parquet(ruta)

print("=== DATOS ALMACENADOS ===")
df_lectura.show(truncate=False)

print("=== TOTAL DE REGISTROS ===")
print(df_lectura.count())

print("=== ESTRUCTURA ===")
df_lectura.printSchema()
```
